```python
class Vendeur:
    def __init__(self, nom, age, salaire, specialite):
        self.nom=nom
        self.age=age
        self.salaire=salaire
        self.specialite=specialite
    def se_presenter(self):
        print(f"je m'appel {self.nom}, j'ai {self.age} ans et specialité {self.specialite}")

    def vendre(self, quantite):
        guain=quantite*400
        print(f"{self.nom} a vendu {quantite} beignet et a encaissé {guain} F.")
        return guain
youssouf=Vendeur("Youssouf", 17, 50000, "chocolat")
youssouf.se_presenter()
youssouf.vendre(5)
```

```python
# programme pour calcules la moyenne des eleves
class Eleve:
    def __init__(self, nom, age, classe):
        self.nom=nom
        self.age=age
        self.classe=classe
        self.note=[] # on lesse case vide pour que l'utilisateur mets ces notes
    
    def se_presenter(self): # pour la presentation
        print(f"je suis {self.nom} en classe {self.classe}")
    
    def ajouter_note(self, note):
        self.note.append(note) # pouur metre la note

    def moyenne(self):
        if len(self.note)==0:
            return 0 # pour donner un moyenne de 0 
        return sum(self.note)/len(self.note)

youssouf=Eleve("Youssouf", 16, "3emme")
youssouf.ajouter_note(18)
youssouf.ajouter_note(15)
youssouf.ajouter_note(12)
youssouf.se_presenter()
print(f"Moyenne de {youssouf.nom} est {youssouf.moyenne()} ")
```

```python
# programme qui affiche la marque la batrie d'un thelephone
class Telephone:
    def __init__(self, nom, marque, batrie):
        self.nom=nom
        self.marque=marque
        self.batrie=batrie
        if 0>batrie or batrie>100:
            self.batrie==0
youssouf=Telephone("Youssouf","OPPO",45)
print(f"bonjour le telephone de {youssouf.nom} est de maque {youssouf.marque} a {youssouf.batrie} %")
```

```python
matieres=[
    ("analyse_3", 3),
    ("topologie_1", 3),
    ("info_de_base_3", 2),
    ("analyse_4", 3),
    ("algebre_lineaire_2", 2),
    ("probabilite_1", 1),
    ("analyse_numerique_1", 1),
    ("mecanique_du_solide", 1),
]
import csv
with open("relever.csv", "w",newline="") as fichier:
    writer=csv.writer(fichier)
    writer.writerow(["matiere","note_cc","examen","coef","moyenne"])

    for matiere, coef in matieres:
        print(f"\n---saisienpour : {matiere} (coef: {coef})---")
        note_cc=float(input("entrer la note du controle continu"))
        while 0<=note_cc and note_cc>=20:
            note_cc=float(input("la note est sur 20"))
        examen=float(input("entrer la note de l'examen"))
        while 0<=examen and examen>=20:
            examen=float(input("la note est sur 20"))
        moyenne=(note_cc*0.25)+(examen*0.75)

        writer.writerow([matiere, note_cc, examen, coef, moyenne])
with open("relever.csv", "r",newline="") as fichier:
    reader=csv.reader(fichier)
    for ligne in reader:
        print(ligne)
```
